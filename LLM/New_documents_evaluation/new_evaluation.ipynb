# Model Output Summary
This notebook summarizes the performance and output count for each model based on the processed constraints CSV.

```python
import pandas as pd
```

```python
csv_path = "project_folder/LLM/New_documents_processed/all_processed_constraints.csv"
df = pd.read_csv(csv_path)
```

```python
# Show available models
models = df[['model_type', 'model_name']].drop_duplicates().reset_index(drop=True)
models
```

```python
# Count outputs for each model
output_counts = df.groupby(['model_type', 'model_name']).size().reset_index(name='output_count')
output_counts
```

```python
# Summarize performance (basic: just output count per model for now)
for _, row in output_counts.iterrows():
    print(f"Model: {row['model_type']} / {row['model_name']} - Outputs: {row['output_count']}")
```
