# Combine Existing Plot Images
This notebook combines the four existing plot images (H1, W1, H2, W2) from the Plots folder into two aligned figures:
- H1 and W1 together (horizontally, minimal gap, with wind speed colorbar)
- H2 and W2 together (horizontally, minimal gap, with wind speed colorbar)

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import matplotlib as mpl
import os
import numpy as np

plots_dir = '/Users/li/Library/CloudStorage/OneDrive-UniversityofMaryland/PhD research/Python/Wind/Digital_Twin/Response_hurricane/Plots'

# Example wind speed range (adjust as needed)
vmin, vmax = 0, 80  # knots

# Use the same colormap and normalization as in the folium HTML (yellow to red, wind speed in m/s)
from matplotlib.colors import LinearSegmentedColormap

# Define color stops as in the HTML (yellow, orange, red, darkred)
colors = ['#ffff00', '#ffae00', '#ff0000', '#8b0000']
cmap_custom = LinearSegmentedColormap.from_list('custom_wind', colors, N=256)

# Set vmin/vmax to match the folium colorbar (e.g., 10 to 39 m/s)
vmin_cb, vmax_cb = 10, 39

# Combine H1 and W1 horizontally with a vertical colorbar in between (make identical to H2/W2 plot)
fig = plt.figure(figsize=(16, 6))
gs = fig.add_gridspec(1, 3, width_ratios=[1, 0.025, 1], wspace=0.0)

# Left image (H1)
ax1 = fig.add_subplot(gs[0, 0])
img1 = mpimg.imread(os.path.join(plots_dir, 'H1.png'))
ax1.imshow(img1)
ax1.axis('off')

# Colorbar in the middle
cax = fig.add_subplot(gs[0, 1])
norm = mpl.colors.Normalize(vmin=vmin_cb, vmax=vmax_cb)
cb2 = mpl.colorbar.ColorbarBase(
    cax, cmap=cmap_custom, norm=norm, orientation='vertical',
    ticks=[10, 20, 30, 39]
)
cb2.set_label('Wind Speed (m/s)', fontsize=14, fontname='Helvetica Neue')
cb2.ax.tick_params(labelsize=12)

# Right image (W1)
ax2 = fig.add_subplot(gs[0, 2])
img2 = mpimg.imread(os.path.join(plots_dir, 'W1.png'))
ax2.imshow(img2)
ax2.axis('off')

plt.subplots_adjust(left=0.01, right=0.99, top=0.88, bottom=0.08, wspace=0.0)
plt.show()

# Combine H2 and W2 horizontally with a vertical colorbar in between (keep as before)
fig = plt.figure(figsize=(16, 6))
gs = fig.add_gridspec(1, 3, width_ratios=[1, 0.025, 1], wspace=0.0)

# Left image (H2)
ax1 = fig.add_subplot(gs[0, 0])
img1 = mpimg.imread(os.path.join(plots_dir, 'H2.png'))
ax1.imshow(img1)
ax1.axis('off')

# Colorbar in the middle
cax = fig.add_subplot(gs[0, 1])
norm = mpl.colors.Normalize(vmin=vmin_cb, vmax=vmax_cb)
cb2 = mpl.colorbar.ColorbarBase(
    cax, cmap=cmap_custom, norm=norm, orientation='vertical',
    ticks=[10, 20, 30, 39]
)
cb2.set_label('Wind Speed (m/s)', fontsize=14, fontname='Helvetica Neue')
cb2.ax.tick_params(labelsize=12)

# Right image (W2)
ax2 = fig.add_subplot(gs[0, 2])
img2 = mpimg.imread(os.path.join(plots_dir, 'W2.png'))
ax2.imshow(img2)
ax2.axis('off')

plt.subplots_adjust(left=0.01, right=0.99, top=0.92, bottom=0.08, wspace=0.0)
plt.show()
```
